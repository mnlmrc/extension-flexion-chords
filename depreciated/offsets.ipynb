```python
import pandas as pd
import EFC_learningfMRI.globals as gl
import os
from math import comb
```

```python
pinfo = pd.read_csv(os.path.join(gl.baseDir, 'participants.tsv'), sep='\t')
pinfo = pinfo[(pinfo.good==1) & (pinfo.sn<114)]
pinfo
```

```python
# parse the dot-separated chord IDs into frozensets of chords
to_set = lambda s: frozenset(int(c) for c in s.split('.'))
fmt    = lambda S: '.'.join(str(c) for c in sorted(S))

trained_sets   = [to_set(s) for s in pinfo.trained]
untrained_sets = [to_set(s) for s in pinfo.untrained]

uniq_trained   = set(trained_sets)
uniq_untrained = set(untrained_sets)

print(f"# unique trained sets   : {len(uniq_trained)} of {comb(8, 4)} possible")
for S in sorted(uniq_trained, key=sorted):
    print('   ', fmt(S))

print(f"\n# unique untrained sets : {len(uniq_untrained)} of {comb(8, 4)} possible")
for S in sorted(uniq_untrained, key=sorted):
    print('   ', fmt(S))

# degree of overlap between the two collections of 4-sets
shared = uniq_trained & uniq_untrained
union  = uniq_trained | uniq_untrained
print(f"\n4-sets appearing as BOTH trained and untrained: {len(shared)}")
for S in sorted(shared, key=sorted):
    print('   ', fmt(S))
print(f"Jaccard overlap of the two collections: {len(shared)}/{len(union)} = {len(shared) / len(union):.2f}")
```
