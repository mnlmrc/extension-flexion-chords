```python
import numpy as np
import pandas as pd
import EFC_learningfMRI.globals as gl
import os
import seaborn as sb
import matplotlib.pyplot as plt
```

```python
H = 'L'
atlas = 'ROI'
rois = gl.rois[atlas]
bold = pd.read_csv(os.path.join(gl.baseDir, 'bold', 'bold_segmented.tsv'), sep='\t')
#bold['signal'] = bold['signal'] - bold.groupby(['sn', 'roi', 'Hem', 'sess'])['signal'].transform('mean')
bold = bold.groupby(['sess', 'chord', 'sn', 'time', 'roi', 'Hem']).mean(numeric_only=True).reset_index()
bold = bold[bold.Hem==H]

sess = 3

fig, axs = plt.subplots(1, 7, sharex=True, sharey=True, figsize=(12, 3))

for r, roi in enumerate(rois):
    ax = axs[r]
    sb.lineplot(ax=ax, data=bold[(bold.roi==roi) & (bold.sess==sess)], x='time', y='signal', hue='chord', 
                palette=['red', 'blue'], hue_order=['trained', 'untrained'], err_kws={'lw': 0}, legend=r==6)
    ax.set_title(roi)

ax.legend(title=None, loc='center left', bbox_to_anchor=(1., .5))

plt.show()
```
